```python
import matplotlib.pyplot as plt

dimensions = [1, 2, 3]
auc = [0.8335337270, 0.8436779916, 0.8439292146]

plt.figure(figsize=(7, 5))

plt.plot(dimensions, auc, marker="o", linewidth=2, markersize=8)

for dim, value in zip(dimensions, auc):
    plt.text(
        dim,
        value + 0.001,
        f"{value:.4f}",
        ha="center",
        fontsize=10,
    )

plt.title("Model MLP fullyprooread")
plt.xlabel("Latent dimension")
plt.ylabel("Test AUC")

plt.xticks(dimensions)
plt.ylim(0.825, 0.850)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
